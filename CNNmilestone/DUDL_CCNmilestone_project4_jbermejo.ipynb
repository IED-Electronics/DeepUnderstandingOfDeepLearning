```python
# import libraries 
import numpy as np 

import torch 
import torch.nn as nn
import torch.nn.functional as F

import torchvision

import matplotlib.pyplot as plt
import matplotlib_inline.backend_inline

from torchsummary import summary

from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import train_test_split

from skimage.draw import line_aa
```

```python
# use gpu 
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

```python
# Global variables 

BATCHSIZE = 64
LR = 0.001
```

# Create a function that generates images (1x30x30) of Gaussian noise plus random vertical or horizontal lines

```python
# Function to create an image with Gaussian noise and a random line
def create_image(vertical=True):
    # Create Gaussian noise
    image = np.random.normal(loc=0.5, scale=0.1, size=(1, 30, 30))
    
    # Randomly decide whether to add a vertical or horizontal line
    if vertical==True:
        # Add a vertical line with random width
        col = np.random.randint(0, 30)
        width = np.random.randint(1, 5)  # Random width between 1 and 4
        image[:, :, max(0, col-width//2):min(30, col+width//2+1)] = 1
    else:
        # Add a horizontal line with random width
        row = np.random.randint(0, 30)
        width = np.random.randint(1, 5)  # Random width between 1 and 4
        image[:, max(0, row-width//2):min(30, row+width//2+1), :] = 1
    
    # Clip values to be in the range [0, 1]
    image = np.clip(image, 0, 1)
    return image

# Generate and visualize the image
image = create_image()
plt.imshow(image[0], cmap='jet')
plt.title('Generated Image')
plt.show()
```

# Create a set of random images

```python
train_images = []
train_labels = []
test_images = []
test_labels = []

# Generate 1000 training/test images with vertical lines
for i in range(1000):
    image = create_image(vertical=True)
    train_images.append(image)
    train_labels.append(1)  # Label 1 for vertical lines

    image = create_image(vertical=True)
    test_images.append(image)
    test_labels.append(1)  # Label 1 for vertical lines

# Generate 1000 training/test images with horizontal lines
for i in range(1000):
    image = create_image(vertical=False)
    train_images.append(image)
    train_labels.append(0)  # Label 0 for horizontal lines

    image = create_image(vertical=False)
    test_images.append(image)
    test_labels.append(0)  # Label 0 for horizontal lines

# Shuffle the training data
train_data = list(zip(train_images, train_labels))
test_data = list(zip(test_images, test_labels))
np.random.shuffle(train_data)
np.random.shuffle(test_data)
train_images, train_labels = zip(*train_data)
test_images, test_labels = zip(*test_data)

# Convert to numpy arrays
train_images = np.array(train_images)
train_labels = np.array(train_labels)
test_images = np.array(test_images)
test_labels = np.array(test_labels)

# Visualise some examples 
fig, axs = plt.subplots(2, 5, figsize=(10, 4))
for i in range(10):
    axs[i//5, i%5].imshow(train_images[i][0], cmap='jet')
    axs[i//5, i%5].set_title(f'Label: {train_labels[i]}')
    axs[i//5, i%5].axis('off')
plt.tight_layout()
plt.show()
```

# Add the data to dataloaders

```python
train_data_tensors = torch.tensor(train_images, dtype=torch.float32)
train_labels_tensors = torch.tensor(train_labels, dtype=torch.long)
test_data_tensors = torch.tensor(test_images, dtype=torch.float32)
test_labels_tensors = torch.tensor(test_labels, dtype=torch.long)

train_dataset = TensorDataset(train_data_tensors, train_labels_tensors)
test_dataset = TensorDataset(test_data_tensors, test_labels_tensors)

train_loader = DataLoader(train_dataset, batch_size=BATCHSIZE, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=test_dataset.tensors[0].shape[0], shuffle=False)
```

# Create model that detects vertical or horizontal lines

```python
def makeTheNet():
    class fmnistnet(nn.Module):
        def __init__(self):
            super().__init__()

            self.ccn1 = nn.Sequential(
                nn.Conv2d(1, 16, kernel_size=3, stride=1, padding=1),
                nn.BatchNorm2d(16),
                nn.MaxPool2d(2),
                nn.LeakyReLU(),
                nn.Dropout(0.1)
                # output: floor((28+2*1-3)/1) + 1 = 28 / 2 = 14
            )

            self.ccn2 = nn.Sequential(
                nn.Conv2d(16, 32, kernel_size=3, stride=1, padding=1),
                nn.BatchNorm2d(32),
                nn.MaxPool2d(2),
                nn.LeakyReLU(),
                nn.Dropout(0.1)
                # output: floor((14+2*1-3)/1) + 1 = 14 / 2 = 7
            )

            self.ccn3 = nn.Sequential(
                nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1),
                nn.BatchNorm2d(64),
                nn.MaxPool2d(2),
                nn.LeakyReLU(),
                nn.Dropout(0.1)
                # output: floor((7+2*1-3)/1) + 1 = 7 / 2 = 3
            )

            self.fc1 = nn.Sequential(
                nn.Linear(64*3*3, 256),
                nn.BatchNorm1d(256),
                nn.LeakyReLU(),
                nn.Dropout(0.3)
            )

            self.fc2 = nn.Sequential(
                nn.Linear(256, 128),
                nn.BatchNorm1d(128),
                nn.LeakyReLU(),
                nn.Dropout(0.3)
            )

            self.fc3 = nn.Sequential(
                nn.Linear(128, 1)
            )

        def forward(self, x):
            self.ccn1_out = self.ccn1(x)
            self.ccn2_out = self.ccn2(self.ccn1_out)
            self.ccn3_out = self.ccn3(self.ccn2_out)
            
            nUnits = self.ccn3_out.shape.numel()/self.ccn3_out.shape[0] # number of features per sample in the batch (numelems/batchsize)
            ccn3_output = self.ccn3_out.view(-1, int(nUnits)) # -1 gets the batchsize automatically, and nUnits is the number of features per sample

            self.fc1_out = self.fc1(ccn3_output)
            self.fc2_out = self.fc2(self.fc1_out)
            self.output = self.fc3(self.fc2_out)

            return self.output
        
    net = fmnistnet()

    lossfun = nn.BCEWithLogitsLoss()

    optimizer = torch.optim.Adam(net.parameters(), lr=LR)

    return net, lossfun, optimizer
```

# Test the model

```python
net, lossfun, optimizer = makeTheNet()
net.to(device)  # Move the model to the same device as the input tensor

X, y = next(iter(train_loader))
X, y = X.to(device), y.to(device)  # Move input and labels to the same device
yHat = net(X)

summary(net, (1, 30, 30), device=str(device))  # Ensure summary uses the correct device
```

# Create a function that trains the model

```python
def function2trainTheModel(epochs=10):
    net, lossfun, optimizer = makeTheNet()

    net.to(device)  # Move the model to the same device as the input tensor

    trainLoss = torch.zeros(epochs)
    testLoss = torch.zeros(epochs)
    trainAcc = torch.zeros(epochs)
    testAcc = torch.zeros(epochs)

    for epoch in range(epochs):
        net.train()

        batchLoss = []
        batchAcc = []

        for X, y in train_loader:
            X, y = X.to(device), y.to(device)

            yHat = net(X)
            loss = lossfun(yHat, y.unsqueeze(dim=1).float())

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            batchLoss.append(loss.item())
            acc = torch.mean(((yHat > 0) == y.unsqueeze(dim=1)).float()).item()
            batchAcc.append(acc)

        trainLoss[epoch] = np.mean(batchLoss)
        trainAcc[epoch] = np.mean(batchAcc)

        net.eval()

        x, y = next(iter(test_loader))
        x, y = x.to(device), y.to(device)

        with torch.no_grad():
            yHat = net(x)
            loss = lossfun(yHat, y.unsqueeze(dim=1).float())

        testLoss[epoch] = loss.item()
        acc = torch.mean(((yHat > 0) == y.unsqueeze(dim=1)).float()).item()
        testAcc[epoch] = acc

    return net, trainLoss, testLoss, trainAcc, testAcc
```

# Test the training function

```python
net, trainLoss, testLoss, trainAcc, testAcc = function2trainTheModel(epochs=2)
```

# Train the model

```python
net, trainLoss, testLoss, trainAcc, testAcc = function2trainTheModel(epochs=10)
```

# Now lets generate some other never seen images and see if it classifies them correctly

```python
def create_image_with_slope(slope):
    # Create a plain background
    image = np.zeros((1, 30, 30))

    # Define the center of the image
    center_x, center_y = 15, 15

    # Calculate the start and end points of the line
    if abs(slope) <= 1:
        # Line is more horizontal
        x_start, x_end = 0, 29
        y_start = int(center_y - slope * center_x)
        y_end = int(center_y + slope * (x_end - center_x))
    else:
        # Line is more vertical
        y_start, y_end = 0, 29
        x_start = int(center_x - (1 / slope) * center_y)
        x_end = int(center_x + (1 / slope) * (y_end - center_y))

    # Clip the start and end points to stay within the image bounds
    x_start = max(0, min(29, x_start))
    x_end = max(0, min(29, x_end))
    y_start = max(0, min(29, y_start))
    y_end = max(0, min(29, y_end))

    # Generate the line using line_aa
    rr, cc, val = line_aa(y_start, x_start, y_end, x_end)

    # Draw the line on the image
    image[0, rr, cc] = val

    # Clip values to be in the range [0, 1]
    image = np.clip(image, 0, 1)
    return image

# Example usage
image = create_image_with_slope(slope=-5)
plt.imshow(image[0], cmap='jet')
plt.show()
```

# Generate test dataset of slopes

```python
slope_images = []
slope_labels = []

for i in range(-64, 64):
    slope_images.append(create_image_with_slope(slope=i/20))
    slope_labels.append(i/20)

fig, axs = plt.subplots(2, 5, figsize=(10, 4))
for i in range(10):
    random_index = np.random.randint(0, len(slope_images))
    axs[i//5, i%5].imshow(slope_images[random_index][0], cmap='jet')
    axs[i//5, i%5].set_title(f'Slope: {slope_labels[random_index]}')
    axs[i//5, i%5].axis('off')
plt.tight_layout()
plt.show()
    
```

# Classify the slopes with our model

```python
slope_images = np.array(slope_images)
slope_labels = np.array(slope_labels)

slope_images_tensors = torch.tensor(slope_images, dtype=torch.float32).to(device)
slope_labels_tensors = torch.tensor(slope_labels, dtype=torch.float32).to(device)

slope_dataset = TensorDataset(slope_images_tensors, slope_labels_tensors)
slope_loader = DataLoader(slope_dataset, batch_size=slope_dataset.tensors[0].size(0), shuffle=False)

results_map = {}
net.eval()

for x, y in slope_loader:
    x, y = x.to(device), y.to(device)

    with torch.no_grad():
        yHat = net(x)
        yHat = torch.sigmoid(yHat)
        yHat = yHat.cpu().numpy()

    for i in range(len(y)):
        results_map[y[i].item()] = yHat[i][0]

for i in range(10):
    print(f"Slope: {list(results_map.keys())[i]} -> Predicted: {list(results_map.values())[i]}")
```
